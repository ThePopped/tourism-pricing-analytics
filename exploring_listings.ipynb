# Exploring All Listings Page

```python
from bs4 import BeautifulSoup
```

```python
with open('listings_chania.html', encoding='utf-8') as f:
    soup = BeautifulSoup(f, 'html.parser')




```

```python
listing_links = soup.find_all(class_="bd77474a8e")
```

```python
len(listing_links)
```

```python
listing_link_urls = []

for listing_link in listing_links:
    listing_link_urls.append(listing_link.get('href'))

listing_link_urls[-10:]



```
